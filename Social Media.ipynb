<font color="#f20a38" size="6"> 🟢 4.Feature Engineering </font>

<font color="#0af2d7" size="5">
Add New Columns Usage_Category Acording to this table <br><br>
| Hours | Category    | <br>
| ----- | ----------- | <br>
| 0–2   | Low User    | <br>
| 3–5   | Medium User | <br>
| >5    | Heavy User  |
</font>

```python
usage_category=np.where(data['daily_social_media_hours']<3,"Low user",
    np.where(data['daily_social_media_hours']<5,"Medium user","Heavy user"))
data['usage_category']=usage_category
usage_count=data['usage_category'].value_counts()
```

<font color="#0af2d7" size="5">
Add New Columns Sleep_Category Acording to this table <br><br>
| Hours | Category    | <br>
| ----- | ----------- | <br>
| 0–4   | Poor Sleep    | <br>
| 5–7   | Normal  Sleep | <br>
| >8   | Healthy Sleep  |
</font>

```python
sleep_category=np.where(data['sleep_hours']<5,"Poor Sleep",
                       np.where(data['sleep_hours']<8,"Normal Sleep","Healthy Sleep"))
data['sleep_category']=sleep_category
```

<font color="#0af2d7" size="5">
Add New Columns Academic_Category Acording to this table <br><br>
| Hours | Category    | <br>
| ----- | ----------- | <br>
| 0–2.5   | Weak        | <br>
| 2.5–3.5   | Average     | <br>
| >3.5    | Excellent   |
</font>

```python
academic_category=np.where(data['academic_performance']<2.5,"Weak",
                          np.where(data['academic_performance']<3.5,"Average","Excellent"))
data['academic_category']=academic_category
```

<font color="#0af2d7" size="5">
Add New Columns Anxiety_Category Acording to this table <br><br>
| Hours | Category    | <br>
| ----- | ----------- | <br>
| 0–4   | Low Risk        | <br>
| 4–7   | Medium Risk     | <br>
| >7    | High Risk |
</font>

```python
anxiety_category=np.where(data['anxiety_level']<4,"Low Risk",
                      np.where(data['anxiety_level']<7,"Medium Risk","High Risk"))
data['anxiety_category']=anxiety_category
```

<font color="#0af2d7" size="5">
Add New Columns Stress_Category Acording to this table <br><br>
| Hours | Category    | <br>
| ----- | ----------- | <br>
| 0–4   | Low Risk        | <br>
| 4–7   | Medium Risk     | <br>
| >7    | High Risk |
</font>

```python
stress_category=np.where(data['stress_level']<4,"Low Risk",
                        np.where(data['stress_level']<7,"Medium Risk","High Risk"))
data['stress_category']=stress_category
```

<font color="#0af2d7" size="5">
Add New Columns Addiction_Category Acording to this table <br><br>
| Hours | Category    | <br>
| ----- | ----------- | <br>
| 0–4   | Low Risk        | <br>
| 4–7   | Medium Risk     | <br>
| >7    | High Risk |
</font>

```python
addiction_category=np.where(data['addiction_level']<4,"Low Risk",
                        np.where(data['addiction_level']<7,"Medium Risk","High Risk"))
data['addiction_category']=addiction_category
```

<font color="yellow" size="5"> 📊 Bar Charts</font>

<font color="#0af2d7" size="5"> Gender Vs Stress </font>
